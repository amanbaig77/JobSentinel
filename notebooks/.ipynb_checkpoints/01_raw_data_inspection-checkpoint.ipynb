```python
df_clean = pd.read_csv("../../data/processed/jobs_cleaned.csv")

print("Shape:", df_clean.shape)
```

```python
missing_analysis = pd.DataFrame({
    "missing_count": df_clean.isnull().sum(),
    "missing_percentage": (
        df_clean.isnull().mean() * 100
    ).round(2)
})

missing_analysis = (
    missing_analysis[missing_analysis["missing_count"] > 0]
    .sort_values("missing_percentage", ascending=False)
)

missing_analysis
```

```python
missing_by_target = pd.DataFrame({
    "legitimate_missing_%": (
        df_clean[df_clean["fraudulent"] == 0]
        .isnull()
        .mean() * 100
    ),
    "fraudulent_missing_%": (
        df_clean[df_clean["fraudulent"] == 1]
        .isnull()
        .mean() * 100
    )
}).round(2)

missing_by_target
```

```python
missing_indicator_columns = [
    "company_profile",
    "salary_range",
    "requirements",
    "benefits",
    "department",
    "required_experience",
    "required_education",
    "industry",
    "function",
    "employment_type",
]

for column in missing_indicator_columns:
    df_clean[f"has_{column}"] = df_clean[column].notna().astype(int)

print("Missing-value indicators created.")
print("New shape:", df_clean.shape)
```

```python
import re

def count_urls(text):
    if pd.isna(text):
        return 0

    return len(
        re.findall(
            r"https?://\S+|www\.\S+",
            str(text),
            flags=re.IGNORECASE
        )
    )
```

```python
for column in text_columns:
    df_clean[f"{column}_url_count"] = (
        df_clean[column].apply(count_urls)
    )

print("URL-count features created.")
```

```python
df_clean.groupby("fraudulent")["description_url_count"].describe()
```

```python
EMAIL_PATTERN = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

def count_emails(text):
    if pd.isna(text):
        return 0

    return len(
        re.findall(
            EMAIL_PATTERN,
            str(text)
        )
    )
```

```python
for column in text_columns:
    df_clean[f"{column}_email_count"] = (
        df_clean[column].apply(count_emails)
    )

print("Email-count features created.")
```

```python
df_clean.groupby("fraudulent")["description_email_count"].describe()
```

```python
PHONE_PATTERN = r"(?<!\d)(?:\+?\d[\d\s().-]{7,}\d)(?!\d)"

def count_phone_numbers(text):
    if pd.isna(text):
        return 0

    return len(
        re.findall(
            PHONE_PATTERN,
            str(text)
        )
    )
```

```python
for column in text_columns:
    df_clean[f"{column}_phone_count"] = (
        df_clean[column].apply(count_phone_numbers)
    )

print("Phone-number features created.")
```

```python
def uppercase_ratio(text):
    if pd.isna(text):
        return 0.0

    text = str(text)

    letters = [char for char in text if char.isalpha()]

    if not letters:
        return 0.0

    uppercase = sum(char.isupper() for char in letters)

    return uppercase / len(letters)
```

```python
for column in text_columns:
    df_clean[f"{column}_uppercase_ratio"] = (
        df_clean[column].apply(uppercase_ratio)
    )

print("Uppercase-ratio features created.")
```

```python
df_clean.groupby("fraudulent")["description_uppercase_ratio"].describe()
```

```python
suspicious_terms = [
    "urgent",
    "immediate",
    "work from home",
    "earn money",
    "easy money",
    "no experience",
    "quick money",
    "guaranteed",
    "investment",
    "cash",
    "wire transfer",
    "bank account",
    "western union",
    "money transfer",
    "commission",
    "telegram",
    "whatsapp",
    "bitcoin",
    "crypto"
]
```

```python
def count_suspicious_terms(text):
    if pd.isna(text):
        return 0

    text = str(text).lower()

    return sum(
        text.count(term)
        for term in suspicious_terms
    )
```

```python
df_clean["suspicious_term_count"] = (
    df_clean["description"]
    .apply(count_suspicious_terms)
)
```

```python
df_clean.groupby("fraudulent")["suspicious_term_count"].describe()
```

```python
from collections import Counter

def get_words(text):
    if pd.isna(text):
        return []

    text = str(text).lower()

    return re.findall(r"\b[a-z]{3,}\b", text)
```

```python
fraud_words = Counter()

for text in df_clean.loc[
    df_clean["fraudulent"] == 1,
    "description"
]:
    fraud_words.update(get_words(text))
```

```python
legit_words = Counter()

for text in df_clean.loc[
    df_clean["fraudulent"] == 0,
    "description"
]:
    legit_words.update(get_words(text))
```

```python
fraud_words.most_common(30)
```

```python
legit_words.most_common(30)
```
