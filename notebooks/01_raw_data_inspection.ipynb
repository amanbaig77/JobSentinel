```python
duplicate_rows = df[df.duplicated(keep=False)].sort_values(
    by=df.columns.tolist()
)

print("Total rows involved in duplicates:", len(duplicate_rows))
```

```python
duplicate_rows["fraudulent"].value_counts()
```

```python
duplicate_groups = (
    duplicate_rows
    .groupby(df.columns.tolist(), dropna=False)["fraudulent"]
    .nunique()
)

print("Duplicate groups with conflicting labels:",
      (duplicate_groups > 1).sum())
```
